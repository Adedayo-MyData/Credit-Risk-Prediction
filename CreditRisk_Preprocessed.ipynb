# CREDIT RISK DATA PREPROCESSING

## Import Relevant Libraries

```python
import pandas as pd
from statsmodels.stats.outliers_influence import variance_inflation_factor
```

## Loading the Raw Data

```python
raw_csv_data = pd.read_csv("credit_risk_raw.csv")
data = raw_csv_data.copy()
```

## Data Preprocessing

```python
pd.options.display.max_columns = None
pd.options.display.max_rows = None
```

```python
data.head()
```

```python
data.info()
```

```python
data.describe()
```

```python
data.isnull().sum()
```

```python
#Removal of duplicates if there is any
data = data.drop_duplicates()
```

```python
data.duplicated().sum()
```

```python
if data['annual_income'].isnull().sum() > 0: 
    data['annual_income']=data['annual_income'].fillna(data['annual_income'].median())
```

```python
if data['total_credit_limit'].isnull().sum() > 0: 
    data['total_credit_limit']=data['total_credit_limit'].fillna(data['total_credit_limit'].median())
```

```python
if data['monthly_debt'].isnull().sum() > 0: 
    data['monthly_debt']=data['monthly_debt'].fillna(data['monthly_debt'].median())
```

```python
if data['loan_amount'].isnull().sum() > 0: 
    data['loan_amount']=data['loan_amount'].fillna(data['monthly_debt'].median())
```

```python
if data['credit_used'].isnull().sum() > 0: 
    data['credit_used']=data['credit_used'].fillna(data['credit_used'].mean())
```

```python
if data['total_delinquencies_12m'].isnull().sum() > 0:
    data['total_delinquencies_12m']=data['total_delinquencies_12m'].fillna(data['total_delinquencies_12m'].median())
```

```python
if data['open_credit_accounts'].isnull().sum() > 0: 
    data['open_credit_accounts']=data['open_credit_accounts'].fillna(data['open_credit_accounts'].mean())
```

```python
if data['credit_history_years'].isnull().sum() > 0:
    data['credit_history_years']=data['credit_history_years'].fillna(data['credit_history_years'].mean())
```

```python
if data['oldest_account_age'].isnull().sum() > 0:
    data['oldest_account_age']=data['oldest_account_age'].fillna(data['oldest_account_age'].mean())
```

```python
if data['late_payments_90d'].isnull().sum() > 0:
    data['late_payments_90d']=data['late_payments_90d'].fillna(data['late_payments_90d'].median())
```

```python
if data['hard_inquiries_last_6m'].isnull().sum() > 0:
    data['hard_inquiries_last_6m']=data['hard_inquiries_last_6m'].fillna(data['hard_inquiries_last_6m'].mean())
```

```python
data.info()
```

```python
data_without_ID=data.drop(['loan_id'],axis=1)
```

```python
data_without_ID.head()
```

Multicollinearity Check

```python
data_without_ID.columns.values
```

```python
variables = data_without_ID[['annual_income', 'monthly_debt', 'loan_amount',
       'total_credit_limit', 'credit_used', 'total_delinquencies_12m',
       'open_credit_accounts', 'credit_history_years',
       'oldest_account_age', 'late_payments_90d',
       'hard_inquiries_last_6m']]

vif = pd.DataFrame()
vif['Feature'] = variables.columns
vif['VIF'] = [variance_inflation_factor(variables.values, i) for i in range(variables.shape[1])]
```

```python
vif
```

## Features Engineering

```python
data_without_ID["debt_to_income"] = data_without_ID["monthly_debt"] / (data_without_ID["annual_income"] / 12)
```

```python
data_without_ID["credit_utilization"] = data_without_ID["credit_used"] / data_without_ID["total_credit_limit"]
```

```python
data_without_ID["loan_to_income"] = data_without_ID["loan_amount"] / data_without_ID["annual_income"]
```

```python
data_without_ID["delinquencies_last_12m"] = data_without_ID["total_delinquencies_12m"]
```

```python
data_without_ID["delinquency_ratio"] = data_without_ID["delinquencies_last_12m"] / data_without_ID["open_credit_accounts"]
```

```python
data_without_ID["utilization_to_income"] = data_without_ID["credit_utilization"] / data_without_ID["annual_income"]
```
